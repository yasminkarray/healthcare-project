```python
import pandas as pd
# Reloading the uploaded file to ensure a fresh start
reclamations_data = pd.ExcelFile(r"C:\Users\HP\Downloads\HEALTHCARE\HEALTHCARE\health_care_database\reclamations.xlsx")
reclamations_df = reclamations_data.parse('Sheet1')

# Displaying the first few rows of the dataset to verify the data is loaded correctly
reclamations_df.head()
```

```python
# General summary of the dataset
eda_summary_reclamations = {
    "Shape": reclamations_df.shape,
    "Missing Values": reclamations_df.isnull().sum(),
    "Column Data Types": reclamations_df.dtypes,
}

# Analyze the distribution of Complaint Types
complaint_type_counts = reclamations_df['Complaint_Type'].value_counts()

# Analyze the distribution of Reclamation Status
status_counts = reclamations_df['Status'].value_counts()

# Analyze Resolution Duration
resolution_duration_stats = reclamations_df['Resolution_Duration_Days'].describe()

# Analyze Satisfaction Impact
satisfaction_impact_counts = reclamations_df['Satisfaction_Impact'].value_counts()

# Display the EDA summary directly
print("Dataset Shape:", eda_summary_reclamations["Shape"])
print("\nMissing Values:\n", eda_summary_reclamations["Missing Values"])
print("\nColumn Data Types:\n", eda_summary_reclamations["Column Data Types"])
```

```python
#  Encodage des colonnes catégoriques "Complaint_Type" et "Status"

# Vérifier les valeurs uniques avant l'encodage
print("\n Valeurs uniques avant encodage:")
print("Complaint_Type:", reclamations_df["Complaint_Type"].unique())
print("Status:", reclamations_df["Status"].unique())

#  Mapping des valeurs catégoriques en valeurs numériques
complaint_type_mapping = {val: idx for idx, val in enumerate(reclamations_df["Complaint_Type"].unique(), start=1)}
status_mapping = {val: idx for idx, val in enumerate(reclamations_df["Status"].unique(), start=1)}

#  Appliquer l'encodage
reclamations_df["Complaint_Type_Encoded"] = reclamations_df["Complaint_Type"].map(complaint_type_mapping)
reclamations_df["Status_Encoded"] = reclamations_df["Status"].map(status_mapping)
```

```python
reclamations_df.head()
```

```python
# Liste des colonnes à supprimer (IDs + colonnes texte inutiles pour le modèle)
colonnes_a_supprimer = ["Reclamation_Date","Reclamation_ID", "Patient_ID","Physician_ID","Admission_ID","ICU_Stay_ID","Complaint_Type","Complaint_Description","Status"]

# Suppression
reclamations_df_clean = reclamations_df.drop(columns=colonnes_a_supprimer)

# Vérification du nouveau DataFrame
print("\n✅ Données prêtes pour la modélisation :")
print(reclamations_df_clean.head())
```

```python
X=reclamations_df_clean.drop('Satisfaction_Impact', axis=1)
Y=reclamations_df_clean['Satisfaction_Impact']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=5)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#KNN
from sklearn.neighbors import KNeighborsClassifier 
KNN=KNeighborsClassifier(n_neighbors=3) 
KNN.fit(X_train,y_train) 
```

```python
print(KNN.score(X_test,y_test))                        
print(KNN.score(X_train,y_train))
```

```python
from sklearn.metrics import classification_report , confusion_matrix

y_pred=KNN.predict(X_test)
print(y_pred)
print(y_test)
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test, y_pred,digits=4))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier

# Définir les hyperparamètres à rechercher
param_grid = {
    'n_neighbors': [2,3,5,],
    'p': [1, 2]  # Mesure de la distance : 1 pour la distance de Manhattan, 2 pour la distance euclidienne
}

# Initialiser le modèle KNN
KNN = KNeighborsClassifier()

# Créer l'objet GridSearchCV
grid_search = GridSearchCV(KNN, param_grid, cv=5)  # cv=5 pour une validation croisée à 5 plis

# Exécuter la recherche sur grille
grid_search.fit(X_train, y_train)  

# Accéder aux meilleurs hyperparamètres et à la meilleure performance
best_params = grid_search.best_params_
best_score = grid_search.best_score_

print("Meilleurs hyperparamètres:", best_params)
print("Meilleure performance :", best_score)
```

```python

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report


# Création et entraînement du modèle de régression logistique
logistic_model = LogisticRegression(max_iter=1000, random_state=42)
logistic_model.fit(X_train, y_train)

# Prédiction sur l'ensemble de test
y_pred = logistic_model.predict(X_test)

# Évaluation du modèle
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Gestion des valeurs manquantes
missing_values = data.isnull().sum()
print("\n--- Valeurs manquantes par colonne ---")
print(missing_values)
```

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split


best_params = {'criterion': 'entropy', 'max_depth': 4}

# Créer le classificateur de base avec les meilleurs paramètres
base_classifier = DecisionTreeClassifier(criterion=best_params['criterion'], max_depth=best_params['max_depth'])

# Créer le modèle de bagging avec le classificateur de base et d'autres paramètres
bagging_model = BaggingClassifier(base_estimator=base_classifier, n_estimators=10, random_state=42, oob_score=True)

# Entraîner le modèle de bagging sur les données d'entraînement
bagging_model.fit(X_train, y_train)

# Calculer et afficher le score du modèle sur les données de test
score = bagging_model.score(X_test, y_test)
print("Score du modèle de bagging:", score)

# Afficher la valeur de l'évaluation OOB
oob_score = bagging_model.oob_score_
print("Score OOB du modèle de bagging avec Random Forest:", oob_score)
```
