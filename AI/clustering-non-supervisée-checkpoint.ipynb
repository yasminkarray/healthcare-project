```python
import pandas as pd

# Charger le fichier Excel
file_path = r"C:\Users\HP\Downloads\HEALTHCARE\HEALTHCARE\health_care_database\PATIENTS.xlsx"
patients_df = pd.read_excel(file_path)

# Afficher les 5 premières lignes
print("Les 5 premières lignes du fichier :")
display(patients_df.head())
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns


#Clustering par comportement de santé
```

```python
# Colonnes liées au profil de santé
columns_to_use = [
    'BMI', 'HadHeartAttack', 'HadStroke', 'HadDiabetes',
    'HadDepressiveDisorder', 'SmokerStatus', 'AlcoholDrinkers',
    'HadAngina','HadAsthma','HadSkinCancer','HadCOPD','HadKidneyDisease'
    ,'HadArthritis', 'DeafOrHardOfHearing','BlindOrVisionDifficulty', 'DifficultyConcentrating',
       'DifficultyWalking', 'DifficultyDressingBathing', 'DifficultyErrands',
]

# Extraire le sous-dataframe
health_df = patients_df[columns_to_use].copy()
```

```python
# Remplacer les "Yes/No" par 1/0
bool_columns = ['HadHeartAttack', 'HadStroke', 'HadDiabetes', 'HadDepressiveDisorder']
for col in bool_columns:
    health_df[col] = health_df[col].map({'Yes': 1, 'No': 0})

# Remplacer les valeurs manquantes
health_df['BMI'].fillna(health_df['BMI'].median(), inplace=True)
for col in bool_columns:
    health_df[col].fillna(0, inplace=True)

# Encoder les colonnes catégoriques
le = LabelEncoder()
for col in ['SmokerStatus', 'AlcoholDrinkers']:
    health_df[col] = le.fit_transform(health_df[col].astype(str))

```

```python
scaler = StandardScaler()
health_scaled = scaler.fit_transform(health_df)
```

```python
pca = PCA(n_components=2)
health_pca = pca.fit_transform(health_scaled)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
clusters = kmeans.fit_predict(health_scaled)
```

```python
# Création d'un DataFrame avec les résultats PCA
pca_df = pd.DataFrame(health_pca, columns=['PCA1', 'PCA2'])
pca_df['Cluster'] = clusters

# Affichage
plt.figure(figsize=(10, 6))
sns.scatterplot(data=pca_df, x='PCA1', y='PCA2', hue='Cluster', palette='Set2', s=60)
plt.title("🔍 Clustering des patients selon leur profil de santé (PCA)")
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(pca_df)  # X : tes données standardisées
    score = silhouette_score(pca_df, labels)
    print(f"Silhouette Score pour k={k} : {score:.4f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(pca_df)

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.title("Clusters visualisés (PCA)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Appliquer KMeans
kmeans = KMeans(n_clusters=2, random_state=42)
labels = kmeans.fit_predict(pca_df)

# Ajouter les labels aux données si tu veux les visualiser ensuite
pca_df['Cluster'] = labels

# Score de silhouette
silhouette = silhouette_score(pca_df.drop(columns=['Cluster']), labels)
print(f"✅ Silhouette Score : {silhouette:.4f}")

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(pca_df.iloc[:, 0], pca_df.iloc[:, 1], c=labels, cmap='viridis', s=50)
plt.title("Clusters visualisés après PCA")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)
plt.show()
```
