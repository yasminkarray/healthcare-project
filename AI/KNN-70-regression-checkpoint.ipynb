```python
import pandas as pd
file_path = "C:/Users/HP/Downloads/reclamations_comple.csv"
reclamations_df = pd.read_csv(file_path)
reclamations_df.head()
```

```python
print("Shape du fichier complet :", reclamations_df.shape)
```

```python
# On concatène le dataframe avec lui-même
df_augmente = pd.concat([reclamations_df, reclamations_df], ignore_index=True)

print("Nombre de lignes après duplication :", df_augmente.shape[0])
```

```python
# Filtrons les lignes à fort impact ou durée de résolution longue
filtre = (reclamations_df["Resolution_Duration_Days"] > 15) | (reclamations_df["Satisfaction_Impact"] < 3)

# On extrait ces lignes
lignes_a_doubler = reclamations_df[filtre]

# On crée un nouveau dataframe augmenté
df_augmente = pd.concat([reclamations_df, lignes_a_doubler], ignore_index=True)

print("Avant :", reclamations_df.shape[0])
print("Après augmentation :", df_augmente.shape[0])
```

```python
print("Shape du fichier complet :", df_augmente.shape)
```

```python
X=df_augmente.drop('Satisfaction_Impact', axis=1)
Y=df_augmente['Satisfaction_Impact']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=5)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#KNN
from sklearn.neighbors import KNeighborsClassifier 
KNN=KNeighborsClassifier(n_neighbors=3) 
KNN.fit(X_train,y_train) 
```

```python
import numpy as np

print(type(X_test))  # Vérifiez le type
if isinstance(X_test, np.ndarray):
    print(X_test.flags)  # Vérifiez les flags du tableau
```

```python
print(KNN.score(X_test,y_test))                        
print(KNN.score(X_train,y_train))
```

```python
from sklearn.metrics import classification_report , confusion_matrix

y_pred=KNN.predict(X_test)
print(y_pred)
print(y_test)
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test, y_pred,digits=4))
```

```python
#  Régression linéaire
model = LinearRegression()
model.fit(X_train, y_train)

#  Prédictions
y_pred = model.predict(X_test)
```

```python
#  Évaluation
mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred, squared=False)
r2 = r2_score(y_test, y_pred)

print(f"MAE : {mae:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R² : {r2:.4f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred, alpha=0.6)
plt.xlabel("Valeurs réelles")
plt.ylabel("Valeurs prédites")
plt.title("Régression : Prédiction vs Réel")
plt.grid(True)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')  # diagonale
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

# Initialisation et entraînement
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Prédictions
y_pred_rf = rf_model.predict(X_test)

# Évaluation
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = mean_squared_error(y_test, y_pred_rf, squared=False)
r2_rf = r2_score(y_test, y_pred_rf)

print(" Random Forest Results :")
print(f"MAE : {mae_rf:.2f}")
print(f"RMSE : {rmse_rf:.2f}")
print(f"R² : {r2_rf:.4f}")
```
